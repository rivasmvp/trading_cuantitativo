```python
# Descarga de datos.

data_sectores = obtener_sectores("5y","1d")
```

```python
import math
import matplotlib.pyplot as plt
import pandas_ta as ta

tickers = data_sectores.columns.levels[0].unique()
n_activos = len(tickers)

if n_activos > 0:
    cols = 2 if n_activos > 1 else 1
    rows_per_activo = 2
    total_rows = math.ceil(n_activos / cols) * rows_per_activo
    
    fig = plt.figure(figsize=(14, 4 * total_rows))
    gs = fig.add_gridspec(total_rows, cols, height_ratios=[3, 1] * (total_rows // 2))
    
    for i, ticker in enumerate(tickers):
        c_idx = i % cols
        r_base = (i // cols) * 2
        
        ax_price = fig.add_subplot(gs[r_base, c_idx])
        ax_rsi = fig.add_subplot(gs[r_base + 1, c_idx], sharex=ax_price)
        
        df_t = data_sectores[ticker].dropna(subset=['Close'])
        if df_t.empty:
            continue
            
        rsi_series = ta.rsi(df_t['Close'], length=14)
        
        ax_price.plot(df_t.index, df_t['Close'], label='Precio', color='#1f77b4', linewidth=1.5)
        ax_price.set_title(f"Sector: {ticker} | RSI: {rsi_series.iloc[-1]:.2f}", fontsize=11, fontweight='bold')
        ax_price.grid(True, linestyle=':', alpha=0.6)
        ax_price.legend(loc='upper left', fontsize=8)
        
        ax_rsi.plot(df_t.index, rsi_series, label='RSI (14)', color='magenta', linewidth=1.2)
        ax_rsi.axhline(70, color='red', linestyle='--', linewidth=0.8, alpha=0.7)
        ax_rsi.axhline(30, color='green', linestyle='--', linewidth=0.8, alpha=0.7)
        ax_rsi.set_ylim(0, 100)
        ax_rsi.grid(True, linestyle=':', alpha=0.6)
        ax_rsi.legend(loc='upper left', fontsize=8)

    plt.tight_layout()
    plt.show()
else:
    print("Sin activos válidos para graficar.")
```
